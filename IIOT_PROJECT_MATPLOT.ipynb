```python
# Step 1: Import libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt=

# Step 2: Read the CSV file
df = pd.read_csv(r"C:\Users\risha\OneDrive\Documents\apple_stock.csv")

# Step 3: Show first 5 rows
print(df.head())

# Step 4: Basic info
print(df.info())

# Step 5: Visualization example
plt.figure(figsize=(10,5))
sns.lineplot(data=df, x="Date", y="Close")
plt.title("Apple Stock Closing Prices Over Time")
plt.show()

```

```python
# === PLOT 2: Bar chart - Volume traded (first 30 rows to keep it readable) ===
# If you prefer last 30 days, use df.tail(30)
sample = df.head(30) if len(df) >= 30 else df
plt.figure(figsize=(12, 5))
plt.bar(sample['Date'], sample['Volume'])
plt.title("Apple Stock Trading Volume (First 30 rows)")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# === PLOT 3: Histogram - Distribution of Closing Prices ===
plt.figure(figsize=(9, 5))
plt.hist(df['Close'].dropna(), bins=20, edgecolor='black')
plt.title("Distribution of Apple Closing Prices")
plt.xlabel("Close Price")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# === PLOT 4: Scatter plot - High vs Low prices ===
plt.figure(figsize=(8, 6))
plt.scatter(df['High'], df['Low'], alpha=0.6)
plt.title("High vs Low Prices of Apple Stock")
plt.xlabel("High Price")
plt.ylabel("Low Price")
plt.grid(True)
plt.tight_layout()
plt.show()
```
